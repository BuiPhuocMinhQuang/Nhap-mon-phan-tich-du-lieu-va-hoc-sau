# 5. Thống kê dữ liệu thiếu cho biến T1 và hiệu chỉnh dữ liệu, lưu ý việc thay thế dữ liệu thiếu sử dụng phương pháp Mean.

```python
# 1. Lập bảng tần số, tần suất để khảo sát dữ liệu thiếu
import pandas as pd

df = pd.read_csv("dulieuxettuyendaihoc.csv")

# Xem các giá trị khác nhau
print("Các giá trị trong T1:")
print(df['T1'].unique())

# Bảng tần số
tan_so = df['T1'].value_counts(dropna=False)

# Bảng tần suất (%)
tan_suat = df['T1'].value_counts(normalize=True, dropna=False) * 100

# Gộp bảng
bang_thong_ke = pd.DataFrame({
    'Tần số': tan_so,
    'Tần suất (%)': tan_suat
})

print("\nBảng tần số - tần suất:")
print(bang_thong_ke)

# Đếm số giá trị thiếu
print("\nSố giá trị thiếu:", df['T1'].isnull().sum())

# 2. Thực hiện thay thế dữ liệu thiếu bằng phương pháp Mean 
# Tính giá trị trung bình
mean_T1 = df['T1'].mean()

print("Giá trị trung bình T1:", mean_T1)

# Thay thế giá trị thiếu bằng mean
df['T1'] = df['T1'].fillna(mean_T1)

# Kiểm tra lại
print("Số giá trị thiếu sau xử lý:", df['T1'].isnull().sum())

# Kiểm tra lại phân bố sau khi xử lý
print(df['T1'].value_counts(dropna=False))
```

# 6. Hãy thực hiện xử lý lần lượt tất cả dữ liệu thiếu cho các biến về điểm số còn lại.

# 7. Tạo các biến TBM1, TBM2, TBM3 tương ứng với trung bình môn của các năm lớp 10, 11 và 12.  
• Công thức tính: TBM = (T*2 + L + H + S + V*2 + X + D + N) / 10

```python
import pandas as pd

df = pd.read_csv("dulieuxettuyendaihoc.csv")

# TBM lớp 10
df['TBM1'] = (df['T1']*2 + df['L1'] + df['H1'] + df['S1'] +
              df['V1']*2 + df['X1'] + df['D1'] + df['N1']) / 10

# TBM lớp 11
df['TBM2'] = (df['T2']*2 + df['L2'] + df['H2'] + df['S2'] +
              df['V2']*2 + df['X2'] + df['D2'] + df['N2']) / 10

# TBM lớp 12
df['TBM3'] = (df['T6']*2 + df['L6'] + df['H6'] + df['S6'] +
              df['V6']*2 + df['X6'] + df['D6'] + df['N6']) / 10

# Xem kết quả
print(df[['TBM1', 'TBM2', 'TBM3']].head())
```

# 8. Tạo các biến xếp loại XL1, XL2 và XL3 dựa trên TBM1,TBM2 và TBM3 cho từng năm lớp 10, 11, 12 như sau: 
• Nhỏ hơn 5.0 xếp loại: yếu (kí hiệu là Y) 
• Từ 5.0 đến dưới 6.5: trung bình (kí hiệu là TB) 
• Từ 6.5 đến dưới 8.0: khá (kí hiệu là K) 
• Từ 8.0 đến dưới 9.0: giỏi (kí hiệu là G) 
• Từ 9.0 trở lên: xuất sắc (kí hiệu là XS) 

```python
# Hàm xếp loại
def xep_loai(tbm):
    if tbm < 5:
        return 'Y'
    elif tbm < 6.5:
        return 'TB'
    elif tbm < 8:
        return 'K'
    elif tbm < 9:
        return 'G'
    else:
        return 'XS'

# Áp dụng cho từng năm
df['XL1'] = df['TBM1'].apply(xep_loai)
df['XL2'] = df['TBM2'].apply(xep_loai)
df['XL3'] = df['TBM3'].apply(xep_loai)

# Xem kết quả
print(df[['TBM1','XL1','TBM2','XL2','TBM3','XL3']].head())
```

# 9. Tạo các biến US_TBM1, US_TBM2 và US_TBM3 để chuyển điểm trung bình các năm lớp 10, 11 và 12 từ thang điểm 10 của Việt Nam sang thang điểm 4 của Mỹ. Sử dụng phương pháp Min-Max Normalization 


```python
import pandas as pd

df = pd.read_csv("dulieuxettuyendaihoc.csv")

df['TBM1'] = (df['T1']*2 + df['L1'] + df['H1'] + df['S1'] + df['V1']*2 + df['X1'] + df['D1'] + df['N1']) / 10
df['TBM2'] = (df['T2']*2 + df['L2'] + df['H2'] + df['S2'] + df['V2']*2 + df['X2'] + df['D2'] + df['N2']) / 10
df['TBM3'] = (df['T6']*2 + df['L6'] + df['H6'] + df['S6'] + df['V6']*2 + df['X6'] + df['D6'] + df['N6']) / 10

# Chuẩn hóa từ thang 10 -> 4
df['US_TBM1'] = df['TBM1'] / 10 * 4
df['US_TBM2'] = df['TBM2'] / 10 * 4
df['US_TBM3'] = df['TBM3'] / 10 * 4

# Xem kết quả
print(df[['TBM1','US_TBM1','TBM2','US_TBM2','TBM3','US_TBM3']].head())
```

# 10. Tạo biến kết quả xét tuyển (kí hiệu là KQXT) nhằm xác định sinh viên đậu (giá trị “1”) và rớt ( giá trị “0”) vào các khối dựa trên điểm DH1, DH2 và DH3 như sau:
• Với khối A, A1 nếu [(DH1*2 + DH2 + DH3)/4] lớn hơn hoặc bằng 5.0 thì đậu, 
ngược lại là rớt 
• Với khối B nếu [(DH1 + DH2*2 + DH3)/4] lớn hơn hoặc bằng 5.0 thì đậu, ngược 
lại là rớt 
• Với khối khác nếu [(DH1+ DH2 + DH3)/3] lớn hơn hoặc bằng 5.0 thì đậu, ngược 
lại là rớt 

```python
import pandas as pd
import numpy as np

df = pd.read_csv("dulieuxettuyendaihoc.csv")

# Tính điểm xét tuyển theo từng khối
df['DIEM_XT'] = np.where(
    df['KT'].isin(['A', 'A1']),
    (df['DH1']*2 + df['DH2'] + df['DH3']) / 4,
    np.where(
        df['KT'] == 'B',
        (df['DH1'] + df['DH2']*2 + df['DH3']) / 4,
        (df['DH1'] + df['DH2'] + df['DH3']) / 3
    )
)

# Xét đậu / rớt
df['KQXT'] = np.where(df['DIEM_XT'] >= 5, 1, 0)

# Xem kết quả
print(df[['KT','DH1','DH2','DH3','DIEM_XT','KQXT']].head())
```

# 1. Trực quan dữ liệu học sinh nữ trên các nhóm XL1, XL2, XL3 dạng unstacked 
Gợi ý - - - - 

Lọc dữ liệu giới tính là nữ 

Oy: Chiều cao biểu đồ cột thể hiển số lượng học sinh theo xếp loại 

Màu sắc thể hiện giá trị xếp loại: [Y, TB, K, G, XS] 

Ox: thể hiện nhóm XL1, XL2 và XL3 

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("processed_dulieuxettuyendaihoc.csv")

# 👉 Tạo TBM
df["TBM1"] = df[["T1","L1","H1","S1","V1","X1","D1","N1"]].mean(axis=1)
df["TBM2"] = df[["T2","L2","H2","S2","V2","X2","D2","N2"]].mean(axis=1)
df["TBM3"] = df[["T3","L3","H3","S3","V3","X3","D3","N3"]].mean(axis=1)

# 👉 Hàm xếp loại
def xeploai(x):
    if x < 5:
        return "Y"
    elif x < 6.5:
        return "TB"
    elif x < 8:
        return "K"
    elif x < 9:
        return "G"
    else:
        return "XS"

# 👉 Tạo XL1, XL2, XL3
df["XL1"] = df["TBM1"].apply(xeploai)
df["XL2"] = df["TBM2"].apply(xeploai)
df["XL3"] = df["TBM3"].apply(xeploai)

# 👉 Lọc nữ
df_female = df[df["GT"] == "F"]

# 👉 Đếm theo từng nhóm (đúng yêu cầu đề)
pivot = pd.DataFrame({
    "XL1": df_female["XL1"].value_counts(),
    "XL2": df_female["XL2"].value_counts(),
    "XL3": df_female["XL3"].value_counts()
}).fillna(0)

# Sắp xếp thứ tự chuẩn
order = ["Y", "TB", "K", "G", "XS"]
pivot = pivot.reindex(order)

print(pivot)

# 👉 Vẽ biểu đồ cột (unstacked)
pivot.plot(kind="bar")

plt.title("Học sinh nữ theo XL1, XL2, XL3")
plt.xlabel("Xếp loại")
plt.ylabel("Số lượng")
plt.xticks(rotation=0)
plt.legend(title="Nhóm")
plt.show()
```

# 1. Hãy mô tả và khảo sát phân phối cho biến T1 
Gợi ý 

Mô tả độ tập trung và phân tán của dữ liệu T1

Kiểm chứng phân phối chuẩn QQ-Plot 

Nhận xét và đánh giá về phân phối của T1

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

df = pd.read_csv("processed_dulieuxettuyendaihoc.csv")

def format_number(x):
    try:
        return str(int(x)) if x == int(x) else f"{x:.2f}"
    except:
        return x

# 📊 DESCRIBE (đã format)
desc = df["T1"].describe().apply(format_number)
print(desc)

# Skewness & Kurtosis
print("Skewness:", format_number(df["T1"].skew()))
print("Kurtosis:", format_number(df["T1"].kurt()))

# 📦 Boxplot
plt.boxplot(df["T1"], vert=False)
plt.title("Boxplot T1")
plt.show()

# 👉 5 số + IQR
Q1 = df["T1"].quantile(0.25)
Q2 = df["T1"].median()
Q3 = df["T1"].quantile(0.75)
IQR = Q3 - Q1

print("Min:", format_number(df["T1"].min()))
print("Q1:", format_number(Q1))
print("Median:", format_number(Q2))
print("Q3:", format_number(Q3))
print("Max:", format_number(df["T1"].max()))
print("IQR:", format_number(IQR))

# 📊 Histogram
df["T1"].plot(kind="hist", bins=10)
plt.title("Histogram T1")
plt.show()

# 📈 QQ Plot
stats.probplot(df["T1"], dist="norm", plot=plt)
plt.title("QQ Plot T1")
plt.show()
```

```python
import pandas as pd
import numpy as np

# ===== B1: Đọc file (fix lỗi csv) =====
df = pd.read_csv("patient_heart_rate.csv", engine='python', on_bad_lines='skip')

# ===== B2: Chuẩn hóa cột =====
df.columns = df.columns.str.strip().str.lower()

print("Columns:", df.columns.tolist())

# ===== B3: Lấy cột huyết áp =====
cols = [c for c in df.columns if c.startswith('m') or c.startswith('f')]

print("Cols BP:", cols)

# ===== 🔥 FIX QUAN TRỌNG =====
# ép về numeric (text -> NaN)
df[cols] = df[cols].apply(pd.to_numeric, errors='coerce')

# ===== B4: Tỷ lệ thiếu =====
print("\nTỷ lệ thiếu (%):")
print(df[cols].isnull().mean() * 100)

# ===== B5: Mean toàn bộ =====
global_mean = df[cols].stack().mean()

# ===== B6: Fill missing =====
def fill_bp(row):
    values = row[cols]

    for i in range(len(cols)):
        if pd.isna(values.iloc[i]):

            # 1. trước + sau
            if i > 0 and i < len(cols)-1:
                if pd.notna(values.iloc[i-1]) and pd.notna(values.iloc[i+1]):
                    values.iloc[i] = (values.iloc[i-1] + values.iloc[i+1]) / 2
                    continue

            # 2. 2 trước
            if i >= 2:
                if pd.notna(values.iloc[i-1]) and pd.notna(values.iloc[i-2]):
                    values.iloc[i] = (values.iloc[i-1] + values.iloc[i-2]) / 2
                    continue

            # 3. 2 sau
            if i <= len(cols)-3:
                if pd.notna(values.iloc[i+1]) and pd.notna(values.iloc[i+2]):
                    values.iloc[i] = (values.iloc[i+1] + values.iloc[i+2]) / 2
                    continue

            # 4. mean người
            person_mean = values.mean()
            if pd.notna(person_mean):
                values.iloc[i] = person_mean
                continue

            # 5. mean giới
            gender = cols[i][0]
            gender_cols = [c for c in cols if c.startswith(gender)]
            gender_mean = row[gender_cols].mean()

            if pd.notna(gender_mean):
                values.iloc[i] = gender_mean
                continue

            # 6. mean toàn bộ
            if pd.notna(global_mean):
                values.iloc[i] = global_mean
            else:
                values.iloc[i] = 120

    row[cols] = values
    return row

# ===== B7: Apply =====
df = df.apply(fill_bp, axis=1)

print("\nSau khi xử lý thiếu:")
print(df[cols].isnull().sum())
```

```python
from torch.utils.data import Dataset, DataLoader 
import torch 
import torch.nn as nn 
import numpy as np
import matplotlib.pyplot as plt 
%matplotlib inline 
device = "cuda" if torch.cuda.is_available() else "cpu" 
from torchvision import datasets 
data_folder = '~/data/FMNIST' # This can be any directory you want to  
# download FMNIST to 
fmnist = datasets.FashionMNIST(data_folder, download=True, train=True) 
tr_images = fmnist.data 
tr_targets = fmnist.targets
```

2. Xây dựng một lớp tìm nạp tập dữ liệu. Chúng ta cần ba hàm __init__, __getitem__ và 
__len__: 

```python
class FMNISTDataset(Dataset): 
    def __init__(self, x, y): 
        x = x.float() 
        x = x.view(-1,28*28) 
        self.x, self.y = x, y  
    def __getitem__(self, ix): 
        x, y = self.x[ix], self.y[ix]  
        return x.to(device), y.to(device) 
    def __len__(self):  
        return len(self.x) 
```

 3. Tạo một hàm tạo DataLoader – trn_dl từ tập dữ liệu – được gọi là FMNISTDataset. Điều này 
sẽ lấy mẫu 32 ảnh ngẫu nhiên  cho kích thước bó: 

```python
def get_data():  
    train = FMNISTDataset(tr_images, tr_targets)  
    trn_dl = DataLoader(train, batch_size=32, shuffle=True) 
    return trn_dl 
```

4. Xác định một mô hình, cũng như hàm sai số và trình tối ưu hóa: 

```python
from torch.optim import SGD 
def get_model(): 
    model = nn.Sequential( 
        nn.Linear(28 * 28, 1000), 
        nn.ReLU(), 
        nn.Linear(1000, 10) 
    ).to(device) 
    loss_fn = nn.CrossEntropyLoss() 
    optimizer = SGD(model.parameters(), lr=1e-2) 
    return model, loss_fn, optimizer 
tr_images = fmnist.data 
tr_targets = fmnist.targets 
```

5. Xác định hàm sẽ huấn luyện tập dữ liệu trên một loạt hình ảnh: 

```python
def train_batch(x, y, model, opt, loss_fn): 
    model.train() # <- let's hold on to this until we reach dropout section 
    # call your model like any python function on your batch of inputs 
    prediction = model(x) 
    # compute loss 
    batch_loss = loss_fn(prediction, y) 
    # based on the forward pass in `model(x)` compute all the gradients of      
    # 'model.parameters()' 
    batch_loss.backward() 
    # apply new-weights = f(old-weights, old-weight-gradients) where  
    # "f" is the optimizer 
    optimizer.step() 
    # Flush gradients memory for next batch of calculations 
    optimizer.zero_grad() 
    return batch_loss.item() 
```

6. Xây dựng hàm tính toán độ chính xác của tập dữ liệu nhất định:

```python
# since there's no need for updating weights, 
# we might as well not compute the gradients. 
# Using this '@' decorator on top of functions 
# will disable gradient computation in the entire function 
@torch.no_grad() 
def accuracy(x, y, model): 
    model.eval() # <- let's wait till we get to dropout section 
    # get the prediction matrix for a tensor of `x` images 
    prediction = model(x) 
    # compute if the location of maximum in each row coincides  
    # with ground truth 
    max_values, argmaxes = prediction.max(-1) 
    is_correct = argmaxes == y 
    return is_correct.cpu().numpy().tolist() 
```

4. Xác định lớp tập dữ liệu (FashionMNIST), các hàm sẽ được sử dụng để huấn luyện trên một

```python
from torchvision import datasets 
import torch 
data_folder = '~/data/FMNIST' # This can be any directory you want to  
# download FMNIST to 
fmnist = datasets.FashionMNIST(data_folder, download=True, train=True) 
tr_images = fmnist.data 
tr_targets = fmnist.targets
val_fmnist = datasets.FashionMNIST(data_folder, download=True, train=False) 
val_images = val_fmnist.data 
val_targets = val_fmnist.targets 
 
import matplotlib.pyplot as plt 
%matplotlib inline 
import numpy as np 
from torch.utils.data import Dataset, DataLoader 
import torch 
import torch.nn as nn 
device = 'cuda' if torch.cuda.is_available() else 'cpu' 
 
#Hingh Learning Rate 
class FMNISTDataset(Dataset): 
    def __init__(self, x, y): 
        x = x.float() 
        x = x.view(-1,28*28) 
        self.x, self.y = x, y  
    def __getitem__(self, ix): 
        x, y = self.x[ix], self.y[ix]  
        return x.to(device), y.to(device) 
    def __len__(self):  
        return len(self.x) 
 
from torch.optim import SGD, Adam 
def get_model(): 
    model = nn.Sequential( 
        nn.Linear(28 * 28, 1000), 
        nn.ReLU(), 
        nn.Linear(1000, 10) 
    ).to(device) 
 
    loss_fn = nn.CrossEntropyLoss() 
    optimizer = Adam(model.parameters(), lr=1e-1) 
    return model, loss_fn, optimizer 
 
def train_batch(x, y, model, opt, loss_fn): 
    model.train() 
    prediction = model(x) 
    batch_loss = loss_fn(prediction, y) 
    batch_loss.backward() 
    optimizer.step() 
    optimizer.zero_grad() 
    return batch_loss.item() 
 
def accuracy(x, y, model): 
    model.eval() 
    # this is the same as @torch.no_grad  
    # at the top of function, only difference 
    # being, grad is not computed in the with scope 
    with torch.no_grad(): 
        prediction = model(x) 
    max_values, argmaxes = prediction.max(-1) 
    is_correct = argmaxes == y 
    return is_correct.cpu().numpy().tolist() 
 
def get_data():  
    train = FMNISTDataset(tr_images, tr_targets)  
    trn_dl = DataLoader(train, batch_size=32, shuffle=True) 
    val = FMNISTDataset(val_images, val_targets)  
    val_dl = DataLoader(val, batch_size=len(val_images), shuffle=False) 
    return trn_dl, val_dl 
 
@torch.no_grad() 
def val_loss(x, y, model): 
    prediction = model(x) 
    val_loss = loss_fn(prediction, y) 
    return val_loss.item() 
 
trn_dl, val_dl = get_data() 
model, loss_fn, optimizer = get_model() 
 
train_losses, train_accuracies = [], [] 
val_losses, val_accuracies = [], [] 
for epoch in range(5): 
    print(epoch) 
    train_epoch_losses, train_epoch_accuracies = [], [] 
    for ix, batch in enumerate(iter(trn_dl)): 
        x, y = batch 
        batch_loss = train_batch(x, y, model, optimizer, loss_fn) 
        train_epoch_losses.append(batch_loss)  
    train_epoch_loss = np.array(train_epoch_losses).mean() 
 
    for ix, batch in enumerate(iter(trn_dl)): 
        x, y = batch 
        is_correct = accuracy(x, y, model) 
        train_epoch_accuracies.extend(is_correct) 
    train_epoch_accuracy = np.mean(train_epoch_accuracies) 
    for ix, batch in enumerate(iter(val_dl)): 
        x, y = batch 
        val_is_correct = accuracy(x, y, model) 
        validation_loss = val_loss(x, y, model) 
    val_epoch_accuracy = np.mean(val_is_correct) 
    train_losses.append(train_epoch_loss) 
    train_accuracies.append(train_epoch_accuracy) 
    val_losses.append(validation_loss) 
    val_accuracies.append(val_epoch_accuracy) 
 
epochs = np.arange(5)+1 
import matplotlib.ticker as mtick 
import matplotlib.pyplot as plt 
import matplotlib.ticker as mticker 
%matplotlib inline 
plt.subplot(211) 
plt.plot(epochs, train_losses, 'bo', label='Training loss') 
plt.plot(epochs, val_losses, 'r', label='Validation loss') 
plt.gca().xaxis.set_major_locator(mticker.MultipleLocator(1)) 
plt.title('Training and validation loss with 0.1 learning rate') 
plt.xlabel('Epochs') 
plt.ylabel('Loss') 
plt.legend() 
plt.grid('off') 
plt.show() 
plt.subplot(212) 
plt.plot(epochs, train_accuracies, 'bo', label='Training accuracy') 
plt.plot(epochs, val_accuracies, 'r', label='Validation accuracy') 
plt.gca().xaxis.set_major_locator(mticker.MultipleLocator(1)) 
plt.title('Training and validation accuracy with 0.1 learning rate') 
plt.xlabel('Epochs') 
plt.ylabel('Accuracy') 
plt.gca().set_yticklabels(['{:.0f}%'.format(x*100) for x in 
plt.gca().get_yticks()])  
plt.legend() 
plt.grid('off') 
plt.show() 
```

# 4. Xây dựng mô hình Deep Neural Network

```python
import torch.nn as nn
import torch.nn.functional as F

class FashionDNN(nn.Module):
    def __init__(self):
        super(FashionDNN, self).__init__()
        
        self.fc1 = nn.Linear(28*28, 256)
        self.fc2 = nn.Linear(256, 128)
        self.fc3 = nn.Linear(128, 64)
        self.fc4 = nn.Linear(64, 10)

    def forward(self, x):
        x = x.view(-1, 28*28)   # Flatten
        
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        x = F.relu(self.fc3(x))
        x = self.fc4(x)
        
        return x

model = FashionDNN()
print(model)
```

# 5. Loss + Optimizer

```python
import torch.optim as optim

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
```

# 6. Huấn luyện mô hình

```python
train_loader = torch.utils.data.DataLoader(
    train_dataset,
    batch_size=64,
    shuffle=True
)

test_loader = torch.utils.data.DataLoader(
    test_dataset,
    batch_size=64,
    shuffle=False
)

epochs = 5

for epoch in range(epochs):
    running_loss = 0.0
    
    for images, labels in train_loader:
        optimizer.zero_grad()
        
        outputs = model(images)
        loss = criterion(outputs, labels)
        
        loss.backward()
        optimizer.step()
        
        running_loss += loss.item()
    
    print(f"Epoch {epoch+1}, Loss: {running_loss/len(train_loader):.4f}")
```

và ‘yellow’ là 1 ví dụ về bigram vì chúng có thể kết hợp lại và có ý nghĩa, vì thế “is yellow” có
thể được tìm thấy trong văn bản. Ví dụ “fast food”, “pay attention”, “good morning”,..,những
bigram này gọi là collocation.
Lọc các bigram sau khi đã loại các stopword và các dấu câu, dùng lệnh sau:

```python
bgrms = nltk.FreqDist(nltk.bigrams(macbeth_filtered2))
bgrms.most_common(15)
```

Ngoài bigram ra, còn có trigram, sự kết hợp của 3 từ, ta dùng lệnh trigrams()

```python
tgrms = nltk.FreqDist(nltk.trigrams (macbeth_filtered2))
tgrms.most_common(10)
```

# 8. Phân tích cảm xúc người dùng
Phân tích cảm xúc là một lĩnh vực nghiên cứu gần đây để đánh giá quan điểm của người dùng về
một chủ đề nào đó. Việc phân tích này dựa trên về các kỹ thuật khác nhau sử dụng phân tích văn
bản, các lĩnh vực trên các mạng xã hội và các diễn đàn.
Nhờ nhận xét và đánh giá của người dùng, các thuật toán phân tích cảm xúc có thể đánh giá mức
độ đánh giá hoặc đánh giá dựa trên các từ khóa nhất định. Mức độ đánh giá được gọi là ý kiến và
có ba giá trị có thể có: positive, neutral hoặc negative. Việc đánh giá ý kiến này do đó trở thành
một dạng phân loại.
Mục đích của ví dụ dưới đây là tìm các từ lặp lại nhiều nhất dạng positive và dạng negative hoặc
các từ khóa liên quan đến một ý kiến nào đó, sử dụng thuật toán Naïve Bayes để phân loại cảm
xúc người dùng dựa vào các đoạn văn bản movie_reviews của họ.
Trước tiên, download các movie review dùng lệnh sau:

```python
nltk.download('movie_reviews')
```

Sau đó xây dựng tập train dựa vào copus trên. Tạo 1 mảng tên là documents. Mảng này chứa cột
đầu tiên là nội dung review của người dùng, và cột thứ 2 là cột đánh giá positive, neutral, or
negative. Sau đó trộn các dòng này ngẫu nhiên.

```python
import random
reviews = nltk.corpus.movie_reviews
documents = [(list(reviews.words(fileid)), category)
for category in reviews.categories()
for fileid in reviews.fileids(category)]
random.shuffle(documents)
```

Xem nội dung review đầu tiên (dòng 0, cột 0)

```python
first_review = ' '.join(documents[0][0])
print(first_review)
```

Xem kết quả review đầu tiên (dòng 0, cột 1)

```python
documents[0][1]
```

Ta cần tạo bảng phân phối tần số các từ trong copus, bảng này cần chuyển sang dạng list, ta dùng hàm list()

```python
all_words = nltk.FreqDist(w.lower() for w in reviews.words())
word_features = list(all_words)
```

Sau đó, bước tiếp theo là xác định một hàm để tính toán các đặc trưng, tức là những từ đủ quan
trọng để thiết lập ý kiến của một review.

```python
def document_features(document, word_features):
   document_words = set(document)
   features = {}
   for word in word_features:
     features['{}'.format(word)] = (word in document_words)
     return features
```

Khi bạn định nghĩa hàm document_features(), bạn tạo 1 tập các documents

```python
featuresets = [(document_features(d,word_features), c) for (d,c) in documents]
len(featuresets)
```

Tạo tập train và tập test: 1500 dòng đầu dùng cho tập train và 500 dòng còn lại dùng cho tập test
để đánh giá độ chính xác của mô hình.

```python
train_set, test_set = featuresets[1500:], featuresets[:500]
classifier = nltk.NaiveBayesClassifier.train(train_set)
```

Dùng thuật toán Naïve Bayes để phân loại, dùng thư viện NLTK. Sau đó tính toán độ chính xác
của thuật toán

```python
train_set, est_set = featuresets[1500:], featuresets[:500]
classifier = nltk.NaiveBayesClassifier.train(train_set)
print(nltk.classify.accuracy(classifier, test_set))
```

Chúng ta đã hoàn tất việc phân tích, dưới đây các từ với trọng số lớn nhất của các review được
đánh giá là positive và negative

```python
classifier.show_most_informative_features(10)
```

3. Kiểm tra stopwords theo từng ngôn ngữ

```python
from nltk.corpus import stopwords

print("English stopwords:")
print(stopwords.words('english')[:20])

print("\nVietnamese (nếu có trong nltk):")
if 'vietnamese' in stopwords.fileids():
    print(stopwords.words('vietnamese'))
else:
    print("NLTK không hỗ trợ Vietnamese stopwords mặc định")
```

4. Loại bỏ stopwords từ một văn bản

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

text = "This is a simple example showing how stopwords are removed from text"
tokens = word_tokenize(text)

filtered = [w for w in tokens if w.lower() not in stopwords.words('english')]

print(filtered)
```

5. Bỏ qua stopwords từ danh sách stopwords

```python
from nltk.corpus import stopwords

words = ["this", "is", "data", "science", "and", "nlp"]

filtered = list(filter(lambda w: w not in stopwords.words('english'), words))

print(filtered)
```

6. WordNet: định nghĩa + ví dụ

```python
from nltk.corpus import wordnet as wn

word = "computer"
synsets = wn.synsets(word)

for syn in synsets[:3]:
    print("Word:", syn.name())
    print("Definition:", syn.definition())
    print("Example:", syn.examples())
    print("-" * 40)
```

7. Tìm từ đồng nghĩa + trái nghĩa

```python
from nltk.corpus import wordnet as wn

word = "good"

synonyms = set()
antonyms = set()

for syn in wn.synsets(word):
    for lemma in syn.lemmas():
        synonyms.add(lemma.name())
        if lemma.antonyms():
            antonyms.add(lemma.antonyms()[0].name())

print("Synonyms:", list(synonyms)[:10])
print("Antonyms:", list(antonyms))
```

8. POS tag set + regex tagger info

9. So sánh similarity giữa 2 danh từ

```python
from nltk.corpus import wordnet as wn

w1 = wn.synset('dog.n.01')
w2 = wn.synset('cat.n.01')

print("Similarity:", w1.path_similarity(w2))
```

10. So sánh similarity giữa 2 động từ

```python
from nltk.corpus import wordnet as wn

v1 = wn.synset('run.v.01')
v2 = wn.synset('walk.v.01')

print("Verb similarity:", v1.path_similarity(v2))
```

11. Thống kê tên nam/nữ

```python
from nltk.corpus import names

male = names.words('male.txt')
female = names.words('female.txt')

print("Male count:", len(male))
print("Female count:", len(female))

print("\n10 male names:", male[:10])
print("10 female names:", female[:10])
```

12. 15 cặp tên nam-nữ ngẫu nhiên

```python
import random
from nltk.corpus import names

male = names.words('male.txt')
female = names.words('female.txt')

pairs = list(zip(male, female))
random.shuffle(pairs)

for i in range(15):
    print(pairs[i])
```

13. Lấy chữ cái cuối + gắn nhãn

```python
from nltk.corpus import names

labeled = []

for name in names.words('male.txt'):
    labeled.append((name[-1], 'male'))

for name in names.words('female.txt'):
    labeled.append((name[-1], 'female'))

print(labeled[:20])
```
