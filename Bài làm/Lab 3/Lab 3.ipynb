# 11. Hãy khảo sát tỉ lệ dữ liệu thiếu trên biến huyết áp. Dữ liệu bị thiếu thì hãy xử lý bằng phương pháp sau 
• Thay thế bằng giá trị trung bình liền trước và liền sau của người đó. Nếu không 
được thì dùng 2) 

• Thay thế bằng giá trị trung bình 2 giá liền trước của người đó. Nếu không được 
thì dùng 3) 

• Thay thế bằng giá trị trung bình 2 giá liền sau của người đó. Nếu không được thì 
dùng 4) 

• Trung bình của các giá trị huyết áp của người đó. Nếu không được thì dùng 5). 

• Trung bình của các giá trị huyết áp của nhóm giới tính. Nếu không được thì dùng 
6) 

• Trung bình của các giá trị dữ liệu. Nếu không được thì thay bằng mức ổn định 
trong y học.   


```python
import pandas as pd
import numpy as np

# ===== B1: Đọc file (fix lỗi csv) =====
df = pd.read_csv("patient_heart_rate.csv", engine='python', on_bad_lines='skip')

# ===== B2: Chuẩn hóa cột =====
df.columns = df.columns.str.strip().str.lower()

print("Columns:", df.columns.tolist())

# ===== B3: Lấy cột huyết áp =====
cols = [c for c in df.columns if c.startswith('m') or c.startswith('f')]

print("Cols BP:", cols)

# ===== 🔥 FIX QUAN TRỌNG =====
# ép về numeric (text -> NaN)
df[cols] = df[cols].apply(pd.to_numeric, errors='coerce')

# ===== B4: Tỷ lệ thiếu =====
print("\nTỷ lệ thiếu (%):")
print(df[cols].isnull().mean() * 100)

# ===== B5: Mean toàn bộ =====
global_mean = df[cols].stack().mean()

# ===== B6: Fill missing =====
def fill_bp(row):
    values = row[cols]

    for i in range(len(cols)):
        if pd.isna(values.iloc[i]):

            # 1. trước + sau
            if i > 0 and i < len(cols)-1:
                if pd.notna(values.iloc[i-1]) and pd.notna(values.iloc[i+1]):
                    values.iloc[i] = (values.iloc[i-1] + values.iloc[i+1]) / 2
                    continue

            # 2. 2 trước
            if i >= 2:
                if pd.notna(values.iloc[i-1]) and pd.notna(values.iloc[i-2]):
                    values.iloc[i] = (values.iloc[i-1] + values.iloc[i-2]) / 2
                    continue

            # 3. 2 sau
            if i <= len(cols)-3:
                if pd.notna(values.iloc[i+1]) and pd.notna(values.iloc[i+2]):
                    values.iloc[i] = (values.iloc[i+1] + values.iloc[i+2]) / 2
                    continue

            # 4. mean người
            person_mean = values.mean()
            if pd.notna(person_mean):
                values.iloc[i] = person_mean
                continue

            # 5. mean giới
            gender = cols[i][0]
            gender_cols = [c for c in cols if c.startswith(gender)]
            gender_mean = row[gender_cols].mean()

            if pd.notna(gender_mean):
                values.iloc[i] = gender_mean
                continue

            # 6. mean toàn bộ
            if pd.notna(global_mean):
                values.iloc[i] = global_mean
            else:
                values.iloc[i] = 120

    row[cols] = values
    return row

# ===== B7: Apply =====
df = df.apply(fill_bp, axis=1)

print("\nSau khi xử lý thiếu:")
print(df[cols].isnull().sum())
```
